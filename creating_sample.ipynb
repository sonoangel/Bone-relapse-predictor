```python
import pandas as pd
import joblib
import os

expr = pd.read_csv('expresion_filtrada.csv', index_col=0)
features = joblib.load('features_list.pkl')

# expr tiene genes en filas — transponemos primero
expr_T = expr.T  # ahora muestras en filas, genes en columnas

# Verificar que los genes existen
genes_ok = [g for g in features if g in expr_T.columns]
print(f'Genes encontrados: {len(genes_ok)} de {len(features)}')

# Crear carpeta si no existe
os.makedirs('bone-relapse-predictor', exist_ok=True)

# Tomar 5 pacientes de muestra
muestra = expr_T[genes_ok].head(5)
muestra.to_csv('bone-relapse-predictor/muestra_test.csv')

print(f'Archivo creado: muestra_test.csv')
print(f'Contiene {len(muestra)} pacientes y {len(genes_ok)} genes')
print(f'Ruta: bone-relapse-predictor/muestra_test.csv')
```
